```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00352/Online%20Retail.xlsx"

df = pd.read_excel(url)

print("=" * 60)
print("DATASET LOADED")
print("=" * 60)

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nColumn Names:")
print(df.columns.tolist())


print("\n" + "=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print(df.info())

print("\nFirst 5 rows:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\n" + "=" * 60)
print("DATA QUALITY CHECK")
print("=" * 60)

negative_quantity = (df['Quantity'] < 0).sum()
zero_negative_price = (df['UnitPrice'] <= 0).sum()

print("Negative Quantity Records:", negative_quantity)
print("Zero/Negative Price Records:", zero_negative_price)

print("\nExamples of Negative Quantity Transactions:")
display(df[df['Quantity'] < 0].head(10))

print("\nExamples of Zero/Negative Price Transactions:")
display(df[df['UnitPrice'] <= 0].head(10))

clean_df = df.copy()


clean_df = clean_df.drop_duplicates()


clean_df = clean_df[clean_df['UnitPrice'] > 0]


clean_df = clean_df[clean_df['Quantity'] > 0]


clean_df['Revenue'] = (
    clean_df['Quantity'] * clean_df['UnitPrice']
)



print("\n" + "=" * 60)
print("CLEANED DATASET")
print("=" * 60)

print("Original Rows:", len(df))
print("Cleaned Rows:", len(clean_df))
print("Columns:", len(clean_df.columns))

print("\nCleaned Dataset Shape:")
print(clean_df.shape)

print("\nRevenue Preview:")
display(
    clean_df[
        ['Quantity', 'UnitPrice', 'Revenue']
    ].head(10)
)



customer_df = clean_df.dropna(
    subset=['CustomerID']
).copy()

print("\n" + "=" * 60)
print("CUSTOMER DATASET")
print("=" * 60)

print("Customer Analysis Rows:", len(customer_df))

print(
    "Customers with valid CustomerID:",
    customer_df['CustomerID'].nunique()
)


total_revenue = clean_df['Revenue'].sum()

total_orders = clean_df['InvoiceNo'].nunique()

unique_products = clean_df['StockCode'].nunique()

unique_customers = clean_df['CustomerID'].nunique()

average_order_value = (
    total_revenue / total_orders
)


print("\n" + "=" * 60)
print("KEY BUSINESS KPIs")
print("=" * 60)

print(f"Total Revenue: £{total_revenue:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Unique Products: {unique_products:,}")
print(f"Unique Customers: {unique_customers:,}")
print(f"Average Order Value: £{average_order_value:,.2f}")




country_revenue = (
    clean_df
    .groupby('Country')['Revenue']
    .sum()
    .sort_values(ascending=False)
)

print("\n" + "=" * 60)
print("TOP 10 COUNTRIES BY REVENUE")
print("=" * 60)

display(
    country_revenue.head(10)
)



plt.figure(figsize=(10, 5))

country_revenue.head(10).plot(
    kind='bar'
)

plt.title("Top 10 Countries by Revenue")
plt.xlabel("Country")
plt.ylabel("Revenue (£)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



clean_df['Month'] = (
    clean_df['InvoiceDate']
    .dt.to_period('M')
)

monthly_revenue = (
    clean_df
    .groupby('Month')['Revenue']
    .sum()
    .sort_index()
)

print("\n" + "=" * 60)
print("MONTHLY REVENUE")
print("=" * 60)

display(monthly_revenue)


plt.figure(figsize=(12, 5))

monthly_revenue.plot()

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (£)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



highest_month = monthly_revenue.idxmax()
highest_revenue = monthly_revenue.max()

lowest_month = monthly_revenue.idxmin()
lowest_revenue = monthly_revenue.min()

print("\n" + "=" * 60)
print("MONTHLY REVENUE INSIGHTS")
print("=" * 60)

print(
    f"Highest Revenue Month: {highest_month}"
)

print(
    f"Highest Revenue: £{highest_revenue:,.2f}"
)

print(
    f"Lowest Revenue Month: {lowest_month}"
)

print(
    f"Lowest Revenue: £{lowest_revenue:,.2f}"
)



top_products = (
    clean_df
    .groupby('Description')['Revenue']
    .sum()
    .sort_values(ascending=False)
)

print("\n" + "=" * 60)
print("TOP 10 PRODUCTS BY REVENUE")
print("=" * 60)

display(
    top_products.head(10)
)



plt.figure(figsize=(10, 6))

top_products.head(10).sort_values().plot(
    kind='barh'
)

plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue (£)")
plt.ylabel("Product")

plt.tight_layout()
plt.show()



customer_revenue = (
    customer_df
    .groupby('CustomerID')['Revenue']
    .sum()
    .sort_values(ascending=False)
)

print("\n" + "=" * 60)
print("TOP 10 CUSTOMERS BY REVENUE")
print("=" * 60)

display(
    customer_revenue.head(10)
)



customer_orders = (
    customer_df
    .groupby('CustomerID')['InvoiceNo']
    .nunique()
    .sort_values(ascending=False)
)

print("\n" + "=" * 60)
print("TOP 10 CUSTOMERS BY NUMBER OF ORDERS")
print("=" * 60)

display(
    customer_orders.head(10)
)



repeat_customers = (
    customer_orders > 1
).sum()

total_customers = customer_orders.count()

repeat_customer_rate = (
    repeat_customers /
    total_customers *
    100
)

print("\n" + "=" * 60)
print("CUSTOMER RETENTION METRICS")
print("=" * 60)

print(
    f"Total Customers: {total_customers:,}"
)

print(
    f"Repeat Customers: {repeat_customers:,}"
)

print(
    f"Repeat Customer Rate: {repeat_customer_rate:.2f}%"
)



average_customer_revenue = (
    customer_revenue.mean()
)

print(
    f"\nAverage Revenue per Customer: "
    f"£{average_customer_revenue:,.2f}"
)



print("\n")
print("=" * 60)
print("PROJECT SUMMARY")
print("=" * 60)

print(f"""
Dataset:
- Original Transactions: {len(df):,}
- Cleaned Transactions: {len(clean_df):,}
- Columns: {len(clean_df.columns)}

Business KPIs:
- Total Revenue: £{total_revenue:,.2f}
- Total Orders: {total_orders:,}
- Unique Products: {unique_products:,}
- Unique Customers: {unique_customers:,}
- Average Order Value: £{average_order_value:,.2f}

Customer KPIs:
- Repeat Customers: {repeat_customers:,}
- Repeat Customer Rate: {repeat_customer_rate:.2f}%
- Average Revenue per Customer: £{average_customer_revenue:,.2f}

Top Revenue Country:
- {country_revenue.index[0]}

Highest Revenue Month:
- {highest_month}

Top Revenue Product:
- {top_products.index[0]}
""")
```

```python
# ============================================================
# SETUP + LOAD DATA + CLEAN DATA + CREATE SQL TABLE
# ============================================================

import pandas as pd
import sqlite3

# 1. Load the real dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00352/Online%20Retail.xlsx"

df = pd.read_excel(url)

print("Original dataset loaded")
print("Original rows:", len(df))


# 2. Create cleaned dataset
clean_df = df.copy()

# Remove duplicate rows
clean_df = clean_df.drop_duplicates()

# Keep valid positive prices
clean_df = clean_df[clean_df['UnitPrice'] > 0]

# Keep normal sales transactions
clean_df = clean_df[clean_df['Quantity'] > 0]

# Create Revenue
clean_df['Revenue'] = (
    clean_df['Quantity'] * clean_df['UnitPrice']
)

print("Cleaned rows:", len(clean_df))


# 3. Create SQL database
conn = sqlite3.connect(":memory:")

# 4. Put cleaned data into SQL table
clean_df.to_sql(
    "sales",
    conn,
    index=False,
    if_exists="replace"
)

print("SQL database created successfully!")
print("SQL table created: sales")

query = """
SELECT *
FROM sales
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT
    SUM(Revenue) AS Total_Revenue
FROM sales;
"""

total_revenue_sql = pd.read_sql_query(query, conn)

print("TOTAL REVENUE")
display(total_revenue_sql)


# 2. Total Orders
query = """
SELECT
    COUNT(DISTINCT InvoiceNo) AS Total_Orders
FROM sales;
"""

total_orders_sql = pd.read_sql_query(query, conn)

print("TOTAL ORDERS")
display(total_orders_sql)


# 3. Revenue by Country
query = """
SELECT
    Country,
    SUM(Revenue) AS Total_Revenue
FROM sales
GROUP BY Country
ORDER BY Total_Revenue DESC
LIMIT 10;
"""

country_revenue_sql = pd.read_sql_query(query, conn)

print("TOP 10 COUNTRIES BY REVENUE")
display(country_revenue_sql)


# 4. Top Products
query = """
SELECT
    Description,
    SUM(Quantity) AS Units_Sold,
    SUM(Revenue) AS Total_Revenue
FROM sales
GROUP BY Description
ORDER BY Total_Revenue DESC
LIMIT 10;
"""

top_products_sql = pd.read_sql_query(query, conn)

print("TOP 10 PRODUCTS BY REVENUE")
display(top_products_sql)
```

```python
# ============================================================
# CUSTOMER ANALYSIS USING SQL
# ============================================================

# 1. Top 10 customers by revenue
query = """
SELECT
    CustomerID,
    SUM(Revenue) AS Total_Revenue
FROM sales
WHERE CustomerID IS NOT NULL
GROUP BY CustomerID
ORDER BY Total_Revenue DESC
LIMIT 10;
"""

top_customers_sql = pd.read_sql_query(query, conn)

print("TOP 10 CUSTOMERS BY REVENUE")
display(top_customers_sql)


# 2. Number of orders per customer
query = """
SELECT
    CustomerID,
    COUNT(DISTINCT InvoiceNo) AS Number_of_Orders
FROM sales
WHERE CustomerID IS NOT NULL
GROUP BY CustomerID
ORDER BY Number_of_Orders DESC
LIMIT 10;
"""

customer_orders_sql = pd.read_sql_query(query, conn)

print("TOP 10 CUSTOMERS BY NUMBER OF ORDERS")
display(customer_orders_sql)


# 3. Average revenue per customer
query = """
SELECT
    AVG(Customer_Revenue) AS Average_Revenue_Per_Customer
FROM
(
    SELECT
        CustomerID,
        SUM(Revenue) AS Customer_Revenue
    FROM sales
    WHERE CustomerID IS NOT NULL
    GROUP BY CustomerID
);
"""

average_customer_sql = pd.read_sql_query(query, conn)

print("AVERAGE REVENUE PER CUSTOMER")
display(average_customer_sql)
```

```python
query = """
SELECT
    COUNT(*) AS Repeat_Customers
FROM
(
    SELECT
        CustomerID,
        COUNT(DISTINCT InvoiceNo) AS Number_of_Orders
    FROM sales
    WHERE CustomerID IS NOT NULL
    GROUP BY CustomerID
    HAVING COUNT(DISTINCT InvoiceNo) > 1
);
"""

repeat_customers_sql = pd.read_sql_query(query, conn)

print("REPEAT CUSTOMERS")
display(repeat_customers_sql)
```
